# PaySim EDA
Goal: understand the data before modelling. End with 3-4 written findings.

```python
import pandas as pd
import matplotlib.pyplot as plt
from fraud.data import load_raw, time_split, summarize_split

df = load_raw()
df.head()
```

## 1. Shape, types, missing values

```python
print(df.shape)
df.info()
df.isna().sum()
```

## 2. Class balance
What accuracy would a model that always predicts 'not fraud' get?

```python
rate = df['isFraud'].mean()
print(f'fraud rate: {rate:.4%}; always-negative accuracy: {1-rate:.4%}')
```

## 3. Fraud by transaction type

```python
df.groupby('type', observed=True)['isFraud'].agg(['count','sum','mean'])
```

## 4. Amount distribution, fraud vs legitimate

```python
df.groupby('isFraud')['amount'].describe()
```

## 5. Fraud over time (`step`)
Is fraud rate stable across the 30 days? This affects the split.

```python
df.groupby('step')['isFraud'].mean().plot(figsize=(10,3), title='fraud rate by step')
plt.show()
```

## 6. Balance patterns (leakage check)
Compare `oldbalanceOrg` / `newbalanceOrig` for fraud vs legitimate TRANSFER and CASH_OUT rows.

```python
sub = df[df['type'].isin(['TRANSFER','CASH_OUT'])]
sub.assign(emptied=(sub['newbalanceOrig']==0)).groupby('isFraud')['emptied'].mean()
```

## 7. Check the default time split

```python
summarize_split(time_split(df))
```
